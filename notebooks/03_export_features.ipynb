# Export the labeled feature table to CSV


```python
import os
import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine

load_dotenv()
engine = create_engine(os.getenv("MYSQL_URL"))

df_final = pd.read_sql("SELECT * FROM extracted_features", engine)
df_final.to_csv("../data/final_yc_extracted_features.csv", index=False, encoding="utf-8-sig")

print("finish")
```
